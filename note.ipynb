```python
import sqlite3, os
db = os.path.join('Dashboard','sorting_dashboard.db')
if not os.path.exists(db):
    print('MISSING_DB')
else:
    conn = sqlite3.connect(db)
    print(f"--------BEFORE RESET")
    rows = conn.execute("SELECT type, value FROM counts ORDER BY type").fetchall()
    for r in rows:
        print(f"{r[0]}:{r[1]}")
    conn.execute("UPDATE counts SET value = 0")
    conn.commit()
    print(f"--------AFTER RESET")
    rows_after = conn.execute("SELECT type, value FROM counts ORDER BY type").fetchall()
    conn.close()
    for r in rows_after:
        print(f"{r[0]}:{r[1]}")
```
